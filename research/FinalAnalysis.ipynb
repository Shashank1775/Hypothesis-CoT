```python
import pandas as pd
import matplotlib.pyplot as plt

# Creating a cleaned DataFrame for GPT-4o data
gpt4o_data = {
    "Dataset": ["AddSub", "ASDiv", "GSM", "MultiArith", "SVAMP", "AQuA"],
    "Normal_CoT": [92.00, 93.66, 91.50, 95.12, 90.24, 82.44],
    "Normal_Standard": [95.50, 88.63, 89.50, 89.50, 89.76, 83.90],
    "Normal_ComplexCoT": [88.00, 91.70, 91.00, 94.63, 90.73, 84.88],
    "Hypothesis_CoT": [93.50, 91.22, 90.00, 99.51, 91.71, 85.37],
    "Hypothesis_Standard": [96.50, 92.68, 91.50, 99.51, 90.73, 83.41],
    "Hypothesis_ComplexCoT": [90.00, 88.29, 88.50, 99.51, 90.24, 82.44],
}

gpt35_data = {
    "Dataset": ["AddSub", "ASDiv", "GSM", "MultiArith", "SVAMP", "AQuA"],
    "Normal_CoT": [87.50, 85.85, 78.00, 95.61, 80.00, 60.49],
    "Normal_Standard": [90.50, 82.93, 78.00, 81.46, 78.05, 61.46],
    "Normal_ComplexCoT": [87.00, 82.93, 73.00, 94.51, 70.73, 60.00],
    "Hypothesis_CoT": [92.50, 86.83, 77.50, 96.59, 80.98, 61.46],
    "Hypothesis_Standard": [91.50, 89.27, 76.00, 95.61, 84.39, 62.44],
    "Hypothesis_ComplexCoT": [87.50, 86.63, 77.00, 96.10, 78.54, 59.02],
}

df_gpt35 = pd.DataFrame(gpt35_data)

df_gpt4o = pd.DataFrame(gpt4o_data)

# Plotting grouped bar chart for GPT-4o
fig, ax = plt.subplots(figsize=(12, 6))
bar_width = 0.12
index = range(len(df_gpt4o))

positions = {
    "Normal_CoT": [i - 2.5 * bar_width for i in index],
    "Normal_Standard": [i - 1.5 * bar_width for i in index],
    "Normal_ComplexCoT": [i - 0.5 * bar_width for i in index],
    "Hypothesis_CoT": [i + 0.5 * bar_width for i in index],
    "Hypothesis_Standard": [i + 1.5 * bar_width for i in index],
    "Hypothesis_ComplexCoT": [i + 2.5 * bar_width for i in index],
}

for label, pos in positions.items():
    ax.bar(pos, df_gpt4o[label], bar_width, label=label.replace("_", " "))

# Formatting
ax.set_xlabel('Dataset')
ax.set_ylabel('Accuracy (%)')
ax.set_title('Prompting Strategies on GPT-4o Across Datasets')
ax.set_xticks(range(len(df_gpt4o)))
ax.set_xticklabels(df_gpt4o["Dataset"], rotation=45)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)
ax.grid(axis='y', linestyle='--', alpha=0.6)

for label, pos in positions.items():
    ax.bar(pos, df_gpt35[label], bar_width, label=label.replace("_", " "))

ax.set_xlabel('Dataset')
ax.set_ylabel('Accuracy (%)')
ax.set_title('Prompting Strategies w/ HFP on GPT-3.5 Turbo Across Datasets')
ax.set_xticks(range(len(df_gpt35)))
ax.set_xticklabels(df_gpt35["Dataset"], rotation=45)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)
ax.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
