```python
input_path = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/AQuA/train.json"       # your original file
output_path = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/AQuA/trainFixed.json"    # new fixed file

with open(input_path, "r", encoding="utf-8") as infile:
    lines = infile.readlines()

# Strip whitespace and filter empty lines
lines = [line.strip() for line in lines if line.strip()]

# Add commas and wrap in brackets
with open(output_path, "w", encoding="utf-8") as outfile:
    outfile.write("[\n")
    for i, line in enumerate(lines):
        outfile.write(line)
        if i < len(lines) - 1:
            outfile.write(",\n")
    outfile.write("\n]")
```

```python
# Path to the training set
trainSet = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/AQuA/trainFixed.json"

# Path to save the sampled prompts
sampledSet = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/AQuA/sampled_train.json"

# Number of prompts to sample
sample_size = 205

# Load the training set
with open(trainSet, 'r') as file:
    data = json.load(file)  # Assuming the file contains a list of prompts

# Randomly sample 205 prompts
sampled_prompts = random.sample(data, sample_size)

# Extract only the required fields
filtered_prompts = [
    {"question": prompt["question"], "options": prompt["options"], "correct": prompt["correct"]}
    for prompt in sampled_prompts
]

# Save the filtered prompts to a new file
with open(sampledSet, 'w') as file:
    json.dump(filtered_prompts, file, indent=4)

print(f"Successfully sampled {sample_size} prompts and saved to {sampledSet}")
```

Nlu - Asdiv

```python
import xml.etree.ElementTree as ET
import random
import json

# Input ASDiv XML path
xml_path = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/nlu-asdiv-dataset/dataset/ASDiv.xml"

# Output sampled JSON path
sampled_path = "/Users/shashanklahoti/Desktop/Coding Projects/Research_LLM_2025/Chain-of-Contrastive-Thought/testingDatasets/ASDIVsampled_train.json"

# Number of problems to sample
sample_size = 205

# Parse XML
tree = ET.parse(xml_path)
root = tree.getroot()

# Navigate to <ProblemSet> and get all <Problem>
problem_set = root.find("ProblemSet")
problems = problem_set.findall("Problem")

# Sample 205 random problems
sampled = random.sample(problems, sample_size)

# Extract and clean required fields
filtered = []
for prob in sampled:
    try:
        entry = {
            "question": prob.findtext("Question").strip(),
            "answer": prob.findtext("Answer").strip(),
            "formula": prob.findtext("Formula").strip(),
            "solution_type": prob.findtext("Solution-Type").strip(),
            "body": prob.findtext("Body").strip(),
            "id": prob.attrib.get("ID"),
            "grade": prob.attrib.get("Grade")
        }
        filtered.append(entry)
    except Exception as e:
        print(f"Skipping problem {prob.attrib.get('ID')} due to error: {e}")

# Save to JSON
with open(sampled_path, "w") as f:
    json.dump(filtered, f, indent=4)

print(f"✅ Successfully sampled {sample_size} problems and saved to {sampled_path}")
```
